```python
import warnings
warnings.filterwarnings('ignore')
import pandas as pd
import numpy as np
import random
import joblib
from sklearn.model_selection import StratifiedKFold
from sklearn.ensemble import ExtraTreesClassifier
from sklearn.linear_model import Ridge
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import (accuracy_score, f1_score, r2_score,
                              mean_absolute_error, classification_report)

# ══════════════════════════════════════════════════
#  1. تحميل البيانات
# ══════════════════════════════════════════════════

texts_orig = df['original_text'].astype(str).values
y_bin      = df['PHQ8_Binary'].values
y_score    = df['PHQ8_Score'].values

print(f"Dataset: {len(df)} rows")
print(f"Class 0 (not depressed): {(y_bin==0).sum()} | Score range: {y_score[y_bin==0].min()}–{y_score[y_bin==0].max()}")
print(f"Class 1 (depressed)    : {(y_bin==1).sum()} | Score range: {y_score[y_bin==1].min()}–{y_score[y_bin==1].max()}\n")

# ══════════════════════════════════════════════════
#  ✅ FIX 1 — Seed ثابت عشان النتايج متتغيرش
# ══════════════════════════════════════════════════
SEED = 42
random.seed(SEED)
np.random.seed(SEED)

# ══════════════════════════════════════════════════
#  دالة الـ Augmentation
# ══════════════════════════════════════════════════
def augment_text(text, n=2):
    words = text.split()
    if len(words) < 2:
        return [text] * n
    augmented = []
    for _ in range(n):
        new_words = [w for w in words if random.random() > 0.15]
        if not new_words:
            new_words = [words[0]]
        augmented.append(' '.join(new_words))
    return augmented

THRESHOLD = 10   # ✅ PHQ-8 threshold الرسمي
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=SEED)
clf_acc, clf_f1, reg_r2, reg_mae = [], [], [], []
reg_acc, reg_f1_th = [], []          # ✅ FIX 2 — threshold metrics للـ regression
all_y_true, all_y_pred = [], []

# ══════════════════════════════════════════════════
#  2. CLASSIFIER CV
# ══════════════════════════════════════════════════
print("=" * 52)
print("  PART 1 — BINARY CLASSIFIER (PHQ8_Binary)")
print("=" * 52)

for fold, (train_idx, test_idx) in enumerate(skf.split(texts_orig, y_bin), 1):

    random.seed(SEED + fold)
    np.random.seed(SEED + fold)

    X_tr_raw = texts_orig[train_idx]
    X_te     = texts_orig[test_idx]
    y_tr_b   = y_bin[train_idx]
    y_te_b   = y_bin[test_idx]

    X_tr_aug  = list(X_tr_raw)
    y_tr_baug = list(y_tr_b)
    for i, label in enumerate(y_tr_b):
        if label == 1:
            nt = augment_text(X_tr_raw[i], n=2)
            X_tr_aug.extend(nt)
            y_tr_baug.extend([1] * len(nt))

    tfidf_clf = TfidfVectorizer(max_features=2000, ngram_range=(1, 2), sublinear_tf=True)
    X_tr_tfidf = tfidf_clf.fit_transform(X_tr_aug)
    X_te_tfidf = tfidf_clf.transform(X_te)

    clf = ExtraTreesClassifier(n_estimators=300, class_weight=None, random_state=SEED)
    clf.fit(X_tr_tfidf, y_tr_baug)

    y_pred_b = clf.predict(X_te_tfidf)
    acc = accuracy_score(y_te_b, y_pred_b)
    f1  = f1_score(y_te_b, y_pred_b)
    clf_acc.append(acc)
    clf_f1.append(f1)
    all_y_true.extend(y_te_b)
    all_y_pred.extend(y_pred_b)
    print(f"  Fold {fold} → Accuracy: {acc*100:.1f}%   F1: {f1:.3f}")

print(f"\n  ✦ Mean Accuracy : {np.mean(clf_acc)*100:.1f}% ± {np.std(clf_acc)*100:.1f}%")
print(f"  ✦ Mean F1 Score : {np.mean(clf_f1):.3f} ± {np.std(clf_f1):.3f}")
print("\n  ── Classification Report (all folds combined) ──")
print(classification_report(
    all_y_true, all_y_pred,
    target_names=['Not Depressed (0)', 'Depressed (1)']
))

# ══════════════════════════════════════════════════
#  3. REGRESSOR CV (على كل الداتا — class 0 و 1)
#     ✅ FIX 2 — بيحسب Threshold Accuracy كمان
# ══════════════════════════════════════════════════
print("=" * 52)
print("  PART 2 — REGRESSOR (PHQ8_Score 0–24, ALL data)")
print(f"  Threshold = {THRESHOLD}  →  score >= {THRESHOLD} = depressed")
print("=" * 52)

for fold, (train_idx, test_idx) in enumerate(skf.split(texts_orig, y_bin), 1):

    random.seed(SEED + fold)
    np.random.seed(SEED + fold)

    X_tr_raw = texts_orig[train_idx]
    X_te     = texts_orig[test_idx]
    y_tr_s   = y_score[train_idx]
    y_te_s   = y_score[test_idx]
    y_te_b   = y_bin[test_idx]

    X_tr_aug  = list(X_tr_raw)
    y_tr_saug = list(y_tr_s)
    for i in range(len(X_tr_raw)):
        nt = augment_text(X_tr_raw[i], n=2)
        X_tr_aug.extend(nt)
        y_tr_saug.extend([y_tr_s[i]] * len(nt))

    tfidf_reg = TfidfVectorizer(max_features=2000, ngram_range=(1, 2), sublinear_tf=True)
    X_tr_tfidf = tfidf_reg.fit_transform(X_tr_aug)
    X_te_tfidf = tfidf_reg.transform(X_te)

    weights = (np.array(y_tr_saug) + 1) ** 2
    reg = Ridge(alpha=1.0)
    reg.fit(X_tr_tfidf, y_tr_saug, sample_weight=weights)

    y_pred_s = np.clip(reg.predict(X_te_tfidf), 0, 24)

    y_pred_b_th = (y_pred_s >= THRESHOLD).astype(int)
    th_acc = accuracy_score(y_te_b, y_pred_b_th)
    th_f1  = f1_score(y_te_b, y_pred_b_th)

    r2  = r2_score(y_te_s, y_pred_s)
    mae = mean_absolute_error(y_te_s, y_pred_s)
    reg_r2.append(r2);       reg_mae.append(mae)
    reg_acc.append(th_acc);  reg_f1_th.append(th_f1)

    print(f"  Fold {fold} → R²: {r2:.3f}  MAE: {mae:.2f}  Acc@{THRESHOLD}: {th_acc*100:.1f}%  F1: {th_f1:.3f}")

print(f"\n  ✦ Mean R²       : {np.mean(reg_r2):.3f} ± {np.std(reg_r2):.3f}")
print(f"  ✦ Mean MAE      : {np.mean(reg_mae):.2f} ± {np.std(reg_mae):.2f}")
print(f"  ✦ Threshold Acc : {np.mean(reg_acc)*100:.1f}% ± {np.std(reg_acc)*100:.1f}%")
print(f"  ✦ Threshold F1  : {np.mean(reg_f1_th):.3f} ± {np.std(reg_f1_th):.3f}")

# ══════════════════════════════════════════════════
#  4. حفظ الـ Pipeline على كل الداتا
# ══════════════════════════════════════════════════
print("\n  Saving final pipeline on full data...")

random.seed(SEED)
np.random.seed(SEED)

# — Classifier —
X_aug_c = list(texts_orig)
y_aug_c = list(y_bin)
for i, label in enumerate(y_bin):
    if label == 1:
        nt = augment_text(texts_orig[i], n=2)
        X_aug_c.extend(nt)
        y_aug_c.extend([1] * len(nt))

tfidf_clf_final = TfidfVectorizer(max_features=2000, ngram_range=(1, 2), sublinear_tf=True)
clf_final = ExtraTreesClassifier(n_estimators=300, class_weight=None, random_state=SEED)
clf_final.fit(tfidf_clf_final.fit_transform(X_aug_c), y_aug_c)

# — Regressor —
X_aug_r = list(texts_orig)
y_aug_r = list(y_score)
for i in range(len(texts_orig)):
    nt = augment_text(texts_orig[i], n=2)
    X_aug_r.extend(nt)
    y_aug_r.extend([y_score[i]] * len(nt))

tfidf_reg_final = TfidfVectorizer(max_features=2000, ngram_range=(1, 2), sublinear_tf=True)
reg_final = Ridge(alpha=1.0)
reg_final.fit(
    tfidf_reg_final.fit_transform(X_aug_r),
    y_aug_r,
    sample_weight=(np.array(y_aug_r) + 1) ** 2
)

joblib.dump({'tfidf': tfidf_clf_final, 'model': clf_final}, 'classifier_pipeline.pkl')
joblib.dump({'tfidf': tfidf_reg_final, 'model': reg_final, 'threshold': THRESHOLD}, 'regressor_pipeline.pkl')

print("  ✦ classifier_pipeline.pkl  — saved")
print("  ✦ regressor_pipeline.pkl   — saved")

# ══════════════════════════════════════════════════
#  5. FINAL SUMMARY
# ══════════════════════════════════════════════════
print("\n" + "=" * 52)
print("  FINAL SUMMARY")
print("=" * 52)
print(f"  [Classifier]  Accuracy : {np.mean(clf_acc)*100:.1f}%  |  F1 : {np.mean(clf_f1):.3f}")
print(f"  [Regressor ]  R²       : {np.mean(reg_r2):.3f}   |  MAE: {np.mean(reg_mae):.2f}")
print(f"  [Reg→Binary]  Acc@{THRESHOLD}   : {np.mean(reg_acc)*100:.1f}%  |  F1 : {np.mean(reg_f1_th):.3f}")
print("=" * 52)
print("\n  NOTE:")
print("  • Seed=42 ثابت — النتايج مش بتتغير مع كل run ✅")
print(f"  • Threshold={THRESHOLD} → score >= {THRESHOLD} = depressed (PHQ-8 standard) ✅")
print("  • Classifier → augment class=1 only (balancing) ✅")
print("  • Regressor  → augment ALL data (full score range 0–24) ✅")
print("  • Both models saved as .pkl — ready for chatbot ✅")
print("=" * 52)
```

```python
import warnings
warnings.filterwarnings('ignore')
import pandas as pd
import numpy as np
import random
import joblib
from sklearn.model_selection import StratifiedKFold
from sklearn.ensemble import ExtraTreesClassifier
from sklearn.linear_model import Ridge
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import (accuracy_score, f1_score, r2_score,
                              mean_absolute_error, classification_report)

# ══════════════════════════════════════════════════
#  1. تحميل البيانات
# ══════════════════════════════════════════════════

texts_orig = df['original_text'].astype(str).values
y_bin      = df['PHQ8_Binary'].values
y_score    = df['PHQ8_Score'].values

print(f"Dataset: {len(df)} rows")
print(f"Class 0 (not depressed): {(y_bin==0).sum()} | Score range: {y_score[y_bin==0].min()}–{y_score[y_bin==0].max()}")
print(f"Class 1 (depressed)    : {(y_bin==1).sum()} | Score range: {y_score[y_bin==1].min()}–{y_score[y_bin==1].max()}\n")

# ══════════════════════════════════════════════════
#  ✅ FIX 1 — Seed ثابت عشان النتايج متتغيرش
# ══════════════════════════════════════════════════
SEED = 42
random.seed(SEED)
np.random.seed(SEED)

# ══════════════════════════════════════════════════
#  دالة الـ Augmentation
# ══════════════════════════════════════════════════
def augment_text(text, n=2):
    words = text.split()
    if len(words) < 2:
        return [text] * n
    augmented = []
    for _ in range(n):
        new_words = [w for w in words if random.random() > 0.15]
        if not new_words:
            new_words = [words[0]]
        augmented.append(' '.join(new_words))
    return augmented

THRESHOLD = 10   # ✅ PHQ-8 threshold الرسمي
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=SEED)
clf_acc, clf_f1, reg_r2, reg_mae = [], [], [], []
reg_acc, reg_f1_th = [], []          # ✅ FIX 2 — threshold metrics للـ regression
all_y_true, all_y_pred = [], []

# ══════════════════════════════════════════════════
#  2. CLASSIFIER CV
# ══════════════════════════════════════════════════
print("=" * 52)
print("  PART 1 — BINARY CLASSIFIER (PHQ8_Binary)")
print("=" * 52)

for fold, (train_idx, test_idx) in enumerate(skf.split(texts_orig, y_bin), 1):

    # ✅ seed ثابت لكل fold
    random.seed(SEED + fold)
    np.random.seed(SEED + fold)

    X_tr_raw = texts_orig[train_idx]
    X_te     = texts_orig[test_idx]
    y_tr_b   = y_bin[train_idx]
    y_te_b   = y_bin[test_idx]

    X_tr_aug  = list(X_tr_raw)
    y_tr_baug = list(y_tr_b)
    for i, label in enumerate(y_tr_b):
        if label == 1:
            nt = augment_text(X_tr_raw[i], n=2)
            X_tr_aug.extend(nt)
            y_tr_baug.extend([1] * len(nt))

    tfidf_clf = TfidfVectorizer(max_features=2000, ngram_range=(1, 2), sublinear_tf=True)
    X_tr_tfidf = tfidf_clf.fit_transform(X_tr_aug)
    X_te_tfidf = tfidf_clf.transform(X_te)

    clf = ExtraTreesClassifier(n_estimators=300, class_weight=None, random_state=SEED)
    clf.fit(X_tr_tfidf, y_tr_baug)

    y_pred_b = clf.predict(X_te_tfidf)
    acc = accuracy_score(y_te_b, y_pred_b)
    f1  = f1_score(y_te_b, y_pred_b)
    clf_acc.append(acc)
    clf_f1.append(f1)
    all_y_true.extend(y_te_b)
    all_y_pred.extend(y_pred_b)
    print(f"  Fold {fold} → Accuracy: {acc*100:.1f}%   F1: {f1:.3f}")

print(f"\n  ✦ Mean Accuracy : {np.mean(clf_acc)*100:.1f}% ± {np.std(clf_acc)*100:.1f}%")
print(f"  ✦ Mean F1 Score : {np.mean(clf_f1):.3f} ± {np.std(clf_f1):.3f}")
print("\n  ── Classification Report (all folds combined) ──")
print(classification_report(
    all_y_true, all_y_pred,
    target_names=['Not Depressed (0)', 'Depressed (1)']
))

# ══════════════════════════════════════════════════
#  3. REGRESSOR CV (على كل الداتا — class 0 و 1)
#     ✅ FIX 2 — بيحسب Threshold Accuracy كمان
# ══════════════════════════════════════════════════
print("=" * 52)
print("  PART 2 — REGRESSOR (PHQ8_Score 0–24, ALL data)")
print(f"  Threshold = {THRESHOLD}  →  score >= {THRESHOLD} = depressed")
print("=" * 52)

for fold, (train_idx, test_idx) in enumerate(skf.split(texts_orig, y_bin), 1):

    # ✅ seed ثابت لكل fold
    random.seed(SEED + fold)
    np.random.seed(SEED + fold)

    X_tr_raw = texts_orig[train_idx]
    X_te     = texts_orig[test_idx]
    y_tr_s   = y_score[train_idx]
    y_te_s   = y_score[test_idx]
    y_te_b   = y_bin[test_idx]

    X_tr_aug  = list(X_tr_raw)
    y_tr_saug = list(y_tr_s)
    for i in range(len(X_tr_raw)):
        nt = augment_text(X_tr_raw[i], n=2)
        X_tr_aug.extend(nt)
        y_tr_saug.extend([y_tr_s[i]] * len(nt))

    tfidf_reg = TfidfVectorizer(max_features=2000, ngram_range=(1, 2), sublinear_tf=True)
    X_tr_tfidf = tfidf_reg.fit_transform(X_tr_aug)
    X_te_tfidf = tfidf_reg.transform(X_te)

    weights = (np.array(y_tr_saug) + 1) ** 2
    reg = Ridge(alpha=1.0)
    reg.fit(X_tr_tfidf, y_tr_saug, sample_weight=weights)

    y_pred_s = np.clip(reg.predict(X_te_tfidf), 0, 24)

    y_pred_b_th = (y_pred_s >= THRESHOLD).astype(int)
    th_acc = accuracy_score(y_te_b, y_pred_b_th)
    th_f1  = f1_score(y_te_b, y_pred_b_th)

    r2  = r2_score(y_te_s, y_pred_s)
    mae = mean_absolute_error(y_te_s, y_pred_s)
    reg_r2.append(r2);       reg_mae.append(mae)
    reg_acc.append(th_acc);  reg_f1_th.append(th_f1)

    print(f"  Fold {fold} → R²: {r2:.3f}  MAE: {mae:.2f}  Acc@{THRESHOLD}: {th_acc*100:.1f}%  F1: {th_f1:.3f}")

print(f"\n  ✦ Mean R²       : {np.mean(reg_r2):.3f} ± {np.std(reg_r2):.3f}")
print(f"  ✦ Mean MAE      : {np.mean(reg_mae):.2f} ± {np.std(reg_mae):.2f}")
print(f"  ✦ Threshold Acc : {np.mean(reg_acc)*100:.1f}% ± {np.std(reg_acc)*100:.1f}%")
print(f"  ✦ Threshold F1  : {np.mean(reg_f1_th):.3f} ± {np.std(reg_f1_th):.3f}")

# ══════════════════════════════════════════════════
#  4. حفظ الـ Pipeline على كل الداتا
# ══════════════════════════════════════════════════
print("\n  Saving final pipeline on full data...")

random.seed(SEED)
np.random.seed(SEED)

# — Classifier —
X_aug_c = list(texts_orig)
y_aug_c = list(y_bin)
for i, label in enumerate(y_bin):
    if label == 1:
        nt = augment_text(texts_orig[i], n=2)
        X_aug_c.extend(nt)
        y_aug_c.extend([1] * len(nt))

tfidf_clf_final = TfidfVectorizer(max_features=2000, ngram_range=(1, 2), sublinear_tf=True)
clf_final = ExtraTreesClassifier(n_estimators=300, class_weight=None, random_state=SEED)
clf_final.fit(tfidf_clf_final.fit_transform(X_aug_c), y_aug_c)

# — Regressor —
X_aug_r = list(texts_orig)
y_aug_r = list(y_score)
for i in range(len(texts_orig)):
    nt = augment_text(texts_orig[i], n=2)
    X_aug_r.extend(nt)
    y_aug_r.extend([y_score[i]] * len(nt))

tfidf_reg_final = TfidfVectorizer(max_features=2000, ngram_range=(1, 2), sublinear_tf=True)
reg_final = Ridge(alpha=1.0)
reg_final.fit(
    tfidf_reg_final.fit_transform(X_aug_r),
    y_aug_r,
    sample_weight=(np.array(y_aug_r) + 1) ** 2
)

joblib.dump({'tfidf': tfidf_clf_final, 'model': clf_final}, 'classifier_pipeline.pkl')
joblib.dump({'tfidf': tfidf_reg_final, 'model': reg_final, 'threshold': THRESHOLD}, 'regressor_pipeline.pkl')

print("  ✦ classifier_pipeline.pkl  — saved")
print("  ✦ regressor_pipeline.pkl   — saved")

# ══════════════════════════════════════════════════
#  5. FINAL SUMMARY
# ══════════════════════════════════════════════════
print("\n" + "=" * 52)
print("  FINAL SUMMARY")
print("=" * 52)
print(f"  [Classifier]  Accuracy : {np.mean(clf_acc)*100:.1f}%  |  F1 : {np.mean(clf_f1):.3f}")
print(f"  [Regressor ]  R²       : {np.mean(reg_r2):.3f}   |  MAE: {np.mean(reg_mae):.2f}")
print(f"  [Reg→Binary]  Acc@{THRESHOLD}   : {np.mean(reg_acc)*100:.1f}%  |  F1 : {np.mean(reg_f1_th):.3f}")
print("=" * 52)
print("\n  NOTE:")
print("  • Seed=42 ثابت — النتايج مش بتتغير مع كل run ✅")
print(f"  • Threshold={THRESHOLD} → score >= {THRESHOLD} = depressed (PHQ-8 standard) ✅")
print("  • Classifier → augment class=1 only (balancing) ✅")
print("  • Regressor  → augment ALL data (full score range 0–24) ✅")
print("  • Both models saved as .pkl — ready for chatbot ✅")
print("=" * 52)
```

```python
import warnings
warnings.filterwarnings('ignore')
import pandas as pd
import numpy as np
import random
import joblib
from sklearn.model_selection import StratifiedKFold
from sklearn.linear_model import Ridge, Lasso, ElasticNet
from sklearn.ensemble import RandomForestRegressor, ExtraTreesRegressor, GradientBoostingRegressor
from sklearn.svm import SVR
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import accuracy_score, f1_score, r2_score, mean_absolute_error
import xgboost as xgb
import lightgbm as lgb



texts_orig = df['original_text'].astype(str).values
y_bin      = df['PHQ8_Binary'].values
y_score    = df['PHQ8_Score'].values

print(f"Dataset: {len(df)} rows | Class 0: {(y_bin==0).sum()} | Class 1: {(y_bin==1).sum()}\n")

SEED      = 42
THRESHOLD =9.5
random.seed(SEED)
np.random.seed(SEED)

def augment_text(text, n=2):
    words = text.split()
    if len(words) < 2: return [text] * n
    augmented = []
    for _ in range(n):
        new_words = [w for w in words if random.random() > 0.15]
        if not new_words: new_words = [words[0]]
        augmented.append(' '.join(new_words))
    return augmented

# ══════════════════════════════════════════════════
#  3. تعريف كل الموديلات
# ══════════════════════════════════════════════════
models = {
    'Ridge'            : Ridge(alpha=1.0),
    'Lasso'            : Lasso(alpha=0.1, max_iter=3000),
    'ElasticNet'       : ElasticNet(alpha=0.1, l1_ratio=0.5, max_iter=3000),
    'RandomForest'     : RandomForestRegressor(n_estimators=200, random_state=SEED, n_jobs=-1),
    'ExtraTrees'       : ExtraTreesRegressor(n_estimators=200, random_state=SEED, n_jobs=-1),
    'GradientBoosting' : GradientBoostingRegressor(n_estimators=100, random_state=SEED),
    'SVR'              : SVR(kernel='linear', C=1.0),
    'XGBoost'          : xgb.XGBRegressor(n_estimators=100, random_state=SEED, verbosity=0, n_jobs=-1),
    'LightGBM'         : lgb.LGBMRegressor(n_estimators=100, random_state=SEED, verbose=-1, n_jobs=-1),
}

# ══════════════════════════════════════════════════
#  4. Cross-Validation لكل موديل
# ══════════════════════════════════════════════════
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=SEED)
results = {}

print("=" * 68)
print(f"  MODEL COMPARISON — Threshold = {THRESHOLD}")
print("=" * 68)

for model_name, model in models.items():
    r2_list, mae_list, acc_list, f1_list = [], [], [], []

    for fold, (train_idx, test_idx) in enumerate(skf.split(texts_orig, y_bin), 1):
        random.seed(SEED + fold)
        np.random.seed(SEED + fold)

        X_tr_raw = texts_orig[train_idx]
        X_te     = texts_orig[test_idx]
        y_tr_s   = y_score[train_idx]
        y_te_s   = y_score[test_idx]
        y_te_b   = y_bin[test_idx]

        # Augmentation على الـ train بس
        X_tr_aug  = list(X_tr_raw)
        y_tr_saug = list(y_tr_s)
        for i in range(len(X_tr_raw)):
            nt = augment_text(X_tr_raw[i], n=2)
            X_tr_aug.extend(nt)
            y_tr_saug.extend([y_tr_s[i]] * len(nt))

        tfidf = TfidfVectorizer(max_features=2000, ngram_range=(1,2), sublinear_tf=True)
        X_tr_tfidf = tfidf.fit_transform(X_tr_aug)
        X_te_tfidf = tfidf.transform(X_te)
        weights = (np.array(y_tr_saug) + 1) ** 2

        # تدريب (SVR مش بيدعم sample_weight)
        if model_name == 'SVR':
            model.fit(X_tr_tfidf, y_tr_saug)
        else:
            model.fit(X_tr_tfidf, y_tr_saug, sample_weight=weights)

        y_pred_s = np.clip(model.predict(X_te_tfidf), 0, 24)
        y_pred_b = (y_pred_s >= THRESHOLD).astype(int)

        r2_list.append(r2_score(y_te_s, y_pred_s))
        mae_list.append(mean_absolute_error(y_te_s, y_pred_s))
        acc_list.append(accuracy_score(y_te_b, y_pred_b))
        f1_list.append(f1_score(y_te_b, y_pred_b))

    results[model_name] = {
        'R2' : np.mean(r2_list),  'R2_std' : np.std(r2_list),
        'MAE': np.mean(mae_list), 'MAE_std': np.std(mae_list),
        'Acc': np.mean(acc_list), 'Acc_std': np.std(acc_list),
        'F1' : np.mean(f1_list),  'F1_std' : np.std(f1_list),
    }
    print(f"  {model_name:<18} R²:{np.mean(r2_list):.3f}  MAE:{np.mean(mae_list):.2f}  Acc:{np.mean(acc_list)*100:.1f}%  F1:{np.mean(f1_list):.3f}")

# ══════════════════════════════════════════════════
#  5. Ranking
# ══════════════════════════════════════════════════
print("\n" + "=" * 68)
print("  RANKING BY ACCURACY")
print("=" * 68)
ranked = sorted(results.items(), key=lambda x: x[1]['Acc'], reverse=True)
for rank, (name, m) in enumerate(ranked, 1):
    print(f"  {rank}. {name:<18} Acc:{m['Acc']*100:.1f}%±{m['Acc_std']*100:.1f}  F1:{m['F1']:.3f}  R²:{m['R2']:.3f}  MAE:{m['MAE']:.2f}")

best_name, best_m = ranked[0]
print(f"\n  ✦ Best Model : {best_name}  →  Acc: {best_m['Acc']*100:.1f}%  F1: {best_m['F1']:.3f}")

# ══════════════════════════════════════════════════
#  6. حفظ أحسن موديل
# ══════════════════════════════════════════════════
print(f"\n  Saving best model ({best_name})...")
random.seed(SEED); np.random.seed(SEED)

best_model = models[best_name]
X_aug_final  = list(texts_orig)
y_aug_final  = list(y_score)
for i in range(len(texts_orig)):
    nt = augment_text(texts_orig[i], n=2)
    X_aug_final.extend(nt)
    y_aug_final.extend([y_score[i]] * len(nt))

tfidf_final = TfidfVectorizer(max_features=2000, ngram_range=(1,2), sublinear_tf=True)
X_final_tfidf = tfidf_final.fit_transform(X_aug_final)
weights_final = (np.array(y_aug_final) + 1) ** 2

if best_name == 'SVR':
    best_model.fit(X_final_tfidf, y_aug_final)
else:
    best_model.fit(X_final_tfidf, y_aug_final, sample_weight=weights_final)

joblib.dump({'tfidf': tfidf_final, 'model': best_model, 'threshold': THRESHOLD}, 'best_regressor_pipeline.pkl')
print(f"  ✦ best_regressor_pipeline.pkl — saved  (threshold={THRESHOLD})")
print("=" * 68)
```
